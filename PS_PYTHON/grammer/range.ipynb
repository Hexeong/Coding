## range

제너레이터의 방식을 활용하는 대표적인 함수로 range()가 있습니다. 활용법은 다음과 같습니다.

```python
print(list(range(5)))
print(range(5))
print(type(range(5)))
for i in range(5):
    print(i, end=' ')
```

하지만 range()의 특별한 특징이 있는데, 바로 lazy loading입니다.
값을 실제로 가지고 있지 않고, 제너레이터로써 가지고 있다가, 값을 접근할 경우 생성되도록 구현되어 있습니다. 다음 예제로 메모리 점유율이 다른 모습을 볼 수 있습니다.

```python
a = [n for n in range(100000)]
b = range(100000)

print(len(a))
print(len(b))
print(len(a) == len(b))

import sys

print(sys.getsizeof(a))
print(sys.getsizeof(b))
```
