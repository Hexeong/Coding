## 제너레이터

제너레이터는 오래된 기능 중 하나로, 루프의 반복 동작을 제어할 수 있는 루틴 형태를 말합니다. 다음의 예시로 빠르게 알아보겠습니다.

```python
def get_natural_number():
    n = 0
    while True:
        n += 1
        yield n

gen = get_natural_number()  # 함수 호출로 제너레이터 생성
print(gen)  # 제너레이터 객체 출력

g = get_natural_number()
for _ in range(0, 100):
    print(next(g))
```

yield 구문을 통해 제너레이터를 리턴할 수 있습니다. 기존의 함수에서 사용하는 return과 달리 yield는 호출 이후에도 값을 리턴한 뒤 맨 끝에 도달할 때까지 계속 함수를 실행하게 됩니다.
물론 다음 코드는 while True 구문으로 종료 조건이 없기에 next()가 계속 호출되면 계속 실행되게 됩니다.
