## 리스트 컴프리핸션

파이선은 map, filter와 같은 함수형 기능을 지원하며 다음과 같은 람다 표현식도 지원합니다.

```python
list(map(lambda x: x + 10, [1,2,3]))
```

리스트 컴프리헨션이란 기존 리스트를 기반으로 새로운 리스트를 만들어내는 구문으로, Python 2.0부터 지원되었습니다. 이는 람다 표현식에 map이나 filter를 섞어서 사용하는 것에 비해 가독성이 훨씬 높습니다.

```python
[n * 2 for n in range(1, 10 + 1) if n % 2 == 1]
```

만약 리스트 컴프리헨션을 사용하지 않을 경우, 코드는 더욱 길어지게 됩니다. 물론 리스트 컴프리헨션이라고 반드시 리스트만 가능한 게 아니라 딕셔너리 등이 가능하도록 2.7부터 추가됐습니다.

```python
original = {"key1" : 1, "key2" : 2}
a = {}
for key, value in original.items():
    a[key] = value
print(a)
```
