## 타입 힌트

파이썬은 대표적인 동적 타이핑 언어임에도 타입을 지정할 수있는 타입 힌트가 PEP 484 문서에 추가됐다. 이 기능은 파이썬 버전 3.5부터 사용할 수 있습다. CPython의 typing.py에는 선언할 수 있으며, 다음과 같은 형태로 타입을 선언할 수 있습니다.

```python
a: str = "1"
b: int = 1
```

예를 들어 기존의 타입 힌트를 사용하지 않는 파이썬 함수는 다음과 같이 정의해 사용해왔습니다.

```python
def fn(a: int) -> bool:
    return a == 1
```

이처럼 타입 힌트를 사용하게 되면 fn()함수의 파라미터 타입과 반환 타입을 알 수 있습니다.
